### Load the IMDB dataset

```python
import pandas as pd
url = "https://raw.githubusercontent.com/Ankit152/IMDB-sentiment-analysis/master/IMDB-Dataset.csv"
df = pd.read_csv(url)
df.head
```

### Inspect the data

```python
print(df.shape)
print(df.isnull().sum())
df.sample(3)
```

### Install NLTK tools

### Clean the text

```python
import re

stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'\d+', '', text)
    tokens = text.split()
    tokens = [w for w in tokens if w not in stop_words]
    return tokens

df['cleaned_tokens'] = df['review'].astype(str).apply(clean_text)
df[['review', 'cleaned_tokens']].head(3)
```

### Original VS Cleaned

```python
for i in range(3):
    print("ORIGINAL:", df['review'].iloc[i][:200], "...")
    print("CLEANED:", df['cleaned_tokens'].iloc[i][:20], "...")
    print("-" * 80)
```
